# 2. BBox size

```python
def compute_bbox_area(row):
    width = row["w"]
    height = row["h"]
    return width * height


def compute_bbox_aspect_ratio(row):
    width = row["w"]
    height = row["h"]
    if height == 0:
        return 0
    return width / height


def compute_bbox_to_image_ratio(row):
    bbox_area = compute_bbox_area(row)
    img_width = 1024
    img_height = 1024

    image_area = img_width * img_height
    return bbox_area / image_area


bbox_df["area"] = bbox_df.apply(compute_bbox_area, axis=1)
bbox_df["aspect_ratio"] = bbox_df.apply(compute_bbox_aspect_ratio, axis=1)
bbox_df["to_image_ratio"] = bbox_df.apply(compute_bbox_to_image_ratio, axis=1)
bbox_df.head()
```

# 3. Spatial Distribution

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


def generate_bbox_heatmap(df, pathology_name, img_size=(1024, 1024)):
    path_df = df[df["finding_label"] == pathology_name]

    heatmap = np.zeros(img_size)

    for _, row in path_df.iterrows():
        x, y, w, h = int(row["x"]), int(row["y"]), int(row["w"]), int(row["h"])

        heatmap[y : y + h, x : x + w] += 1

    return heatmap
```

```python
pathologies = bbox_df["finding_label"].unique()
num_plots = len(pathologies)

cols = 4
rows = math.ceil(num_plots / cols)
fig, axes = plt.subplots(rows, cols, figsize=(cols * 5, rows * 5))
axes = axes.flatten()

for i, pathology in enumerate(pathologies):
    heatmap = generate_bbox_heatmap(bbox_df, pathology)

    sns.heatmap(heatmap, cmap="Reds", cbar=False, ax=axes[i])

    axes[i].set_title(f"Pathology: {pathology}", fontsize=14)
    axes[i].axis("off")

for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.suptitle("Spatial Distribution of Bounding Boxes across Pathologies", fontsize=20, y=1.02)
plt.show()
```

```python
def calculate_heatmap_center_of_gravity(heatmap):
    total_mass = np.sum(heatmap)

    if total_mass == 0:
        return (0, 0)

    x_coords = np.arange(heatmap.shape[1])
    y_coords = np.arange(heatmap.shape[0])

    x_center_of_gravity = np.sum(np.sum(heatmap, axis=0) * x_coords) / total_mass
    y_center_of_gravity = np.sum(np.sum(heatmap, axis=1) * y_coords) / total_mass

    return (x_center_of_gravity, y_center_of_gravity)
```

```python
gravity_points = {}
for pathology in bbox_df["finding_label"].unique():
    heatmap = generate_bbox_heatmap(bbox_df, pathology)
    center_of_gravity = calculate_heatmap_center_of_gravity(heatmap)
    print(f"Pathology: {pathology}, Center of Gravity: {center_of_gravity}")
    gravity_points[pathology] = center_of_gravity
```

```python
pathologies = bbox_df["finding_label"].unique()
num_plots = len(pathologies)

cols = 4
rows = math.ceil(num_plots / cols)
fig, axes = plt.subplots(rows, cols, figsize=(cols * 5, rows * 5))
axes = axes.flatten()

for i, pathology in enumerate(pathologies):
    heatmap = generate_bbox_heatmap(bbox_df, pathology)

    sns.heatmap(heatmap, cmap="Reds", cbar=False, ax=axes[i])
    axes[i].scatter(gravity_points[pathology][0], gravity_points[pathology][1], color="blue", marker="X", s=100)

    axes[i].set_title(f"Pathology: {pathology}", fontsize=14)
    axes[i].axis("off")

for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.suptitle("Spatial Distribution of Bounding Boxes across Pathologies", fontsize=20, y=1.02)
plt.legend(["Center of Gravity"], loc="lower right")
plt.show()
```

```python
def to_polish(label):
    mapping = {
        "Atelectasis": "Niedodma",
        "Cardiomegaly": "Kardiomegalia",
        "Effusion": "Wysięk",
        "Infiltrate": "Naciek",
        "Mass": "Masa",
        "Nodule": "Guzek",
        "Pneumonia": "Zapalenie płuc",
        "Pneumothorax": "Odma opłucnowa",
        "Consolidation": "Konsolidacja",
        "Edema": "Obrzęk",
        "Emphysema": "Emphysema",
        "Fibrosis": "Fibrosis",
        "Pleural_Thickening": "Grubienie opłucnej",
        "Hernia": "Hernia",
    }
    return mapping.get(label, label)

def show_image_with_bbox(df, image_index):
    image_path = f"data/01_raw/images/{image_index}"
    image = Image.open(image_path).convert("RGB")
    finding_label = df[df["image_index"] == image_index]["finding_label"].iloc[0]
    fig, ax = plt.subplots(1, figsize=(12, 12))
    ax.imshow(image)

    bboxes = df[df["image_index"] == image_index]

    for _, row in bboxes.iterrows():
        x, y, w, h = int(row["x"]), int(row["y"]), int(row["w"]), int(row["h"])
        rect = plt.Rectangle((x, y), w, h, edgecolor="red", facecolor="none", linewidth=2)
        ax.add_patch(rect)

    plt.title(f"{to_polish(finding_label)}", fontsize=14)
    plt.axis("off")
    plt.show()
```
