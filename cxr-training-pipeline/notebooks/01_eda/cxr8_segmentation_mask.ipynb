```python
def get_combined_mask(idx: int) -> np.ndarray:
    left_lung_mask = rle_to_mask(segmentation_masks["Left Lung"][idx], 1024, 1024)
    right_lung_mask = rle_to_mask(segmentation_masks["Right Lung"][idx], 1024, 1024)
    heart_mask = rle_to_mask(segmentation_masks["Heart"][idx], 1024, 1024)
    combined_mask = np.maximum.reduce([left_lung_mask, right_lung_mask, heart_mask])
    return combined_mask
```

```python
def plot_mask_overlay(idx: int) -> None:

    combined_mask = get_combined_mask(idx)

    image_index = segmentation_masks.loc[idx, "Image Index"]
    img_path = os.path.join("data", "01_raw", "images", image_index)

    plt.figure(figsize=(6, 6))
    img = plt.imread(img_path)
    plt.imshow(img, cmap="gray")
    plt.imshow(combined_mask, cmap="pink_r", alpha=0.5)
    plt.axis("off")
    plt.show()
```

```python
def plot_bbox_overlay(idx: int) -> None:
    combined_mask = get_combined_mask(idx)
    bbox_coords = get_bbox_coordinates(combined_mask)

    if bbox_coords is None:
        print("No mask present, skipping bounding box overlay.")
        return

    x_min, y_min, x_max, y_max = bbox_coords
    image_index = segmentation_masks.loc[idx, "Image Index"]
    img_path = os.path.join("data", "01_raw", "images", image_index)

    plt.figure(figsize=(6, 6))
    img = plt.imread(img_path)
    plt.imshow(img, cmap="gray")
    rect = plt.Rectangle((x_min, y_min), x_max - x_min, y_max - y_min,
                         edgecolor='red', facecolor='none', linewidth=2)
    plt.gca().add_patch(rect)
    plt.title("Bounding Box Overlay")
    plt.axis("off")
    plt.show()
```

```python
plot_bbox_overlay(1251)
```
