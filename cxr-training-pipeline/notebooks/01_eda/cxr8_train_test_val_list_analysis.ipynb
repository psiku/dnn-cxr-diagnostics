# 1. Leakage Detection

```python
def calculate_patient_ids(list: list[str], df: pd.DataFrame) -> set[str]:
    patient_ids = []
    for image_idx in tqdm(list, desc="Calculating patient IDs"):
        patient_id = df[df['image_index'] == image_idx]['patient_id'].values[0]
        patient_ids.append(patient_id)
    return set(patient_ids)

train_val_patient_ids = calculate_patient_ids(train_val_list, metadata)
test_patient_ids = calculate_patient_ids(test_list, metadata)

intersection = train_val_patient_ids.intersection(test_patient_ids)

print(f"Number of unique patient IDs in train/val set: {len(train_val_patient_ids)}")
print(f"Number of unique patient IDs in test set: {len(test_patient_ids)}")
print(f"Number of unique patient IDs in both sets: {len(intersection)}")
```

# 2. Labels distribution

```python
def count_labels_in_split(split_list: list[str], df: pd.DataFrame) -> pd.Series:
    unique_labels = df['finding_labels'].str.split('|').explode().unique()
    pathologies = {}
    for img_idx in tqdm(split_list, desc="Counting labels in split"):
        labels = df[df['image_index'] == img_idx]['finding_labels'].values[0]
        for label in labels.split('|'):
            if label not in pathologies:
                pathologies[label] = 0
            pathologies[label] += 1
    return pd.Series(pathologies)
```

### 2.1 train_val

```python
train_val_label_counts = count_labels_in_split(train_val_list, metadata)
```

```python
train_val_label_counts.sort_values(ascending=False).plot(kind='bar', figsize=(12, 6), title='Label Distribution in Train/Val Set')
plt.xlabel('Pathology')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
for name, count in train_val_label_counts.sort_values(ascending=False).items():
    print(f"{name}: {count/len(train_val_list) * 100:.2f}%")
```
