### Load images

```python
IMAGES_FOLDER = r"data/01_raw/images"

def _get_random_image():
    import random

    images = os.listdir(IMAGES_FOLDER)
    return os.path.join(IMAGES_FOLDER, random.choice(images))
```

```python
from torchvision import transforms
def _get_rotation_augmentation():

    return transforms.RandomRotation(degrees=10)

def _get_translation_augmentation():

    return transforms.RandomAffine(degrees=0, translate=(0.05, 0.05))

def _get_random_scale_augmentation():

    return transforms.RandomResizedCrop(size=(224, 224), scale=(0.95, 1.05))
```

### Augmentations examples

```python
import numpy as np
import matplotlib.pyplot as plt
image = _get_random_image()
print(image)

image_as_np = np.array(plt.imread(image))
```

```python
image_pil = transforms.ToPILImage()(image_as_np)

rotated_image = _get_rotation_augmentation()(image_pil)
translated_image = _get_translation_augmentation()(image_pil)
scaled_image = _get_random_scale_augmentation()(image_pil)

fig, axes = plt.subplots(1, 4, figsize=(20, 5))

axes[0].imshow(image_as_np, cmap="gray")
axes[0].set_title("Oryginalne zdjęcie")

axes[1].imshow(rotated_image, cmap="gray")
axes[1].set_title("Po rotacji")

axes[2].imshow(translated_image, cmap="gray")
axes[2].set_title("Po translacji")

axes[3].imshow(scaled_image, cmap="gray")
axes[3].set_title("Po skalowaniu")

for ax in axes:
    ax.axis("off")

plt.tight_layout()
```
