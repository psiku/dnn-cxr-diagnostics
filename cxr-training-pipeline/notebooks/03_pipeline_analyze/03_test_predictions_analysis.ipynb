### Testing model with given thresholds

```python
def load_model(checkpoint_path, model_type, transition_dim, num_classes=14):
    model = ChestXRayClassifier(backbone_name=model_type,
                                pretrained=True,
                                grayscale=True,
                                num_classes=num_classes,
                                transition_dim=transition_dim)
    checkpoint = torch.load(checkpoint_path, map_location=torch.device("cpu"), weights_only=False)

    model_state_dict = checkpoint["state_dict"]

    model_state_dict = {
        k.replace("model.", "", 1): v
        for k, v in model_state_dict.items()
        if not k.startswith("thresholds") and "thresholds" not in k
    }
    model.load_state_dict(model_state_dict)

    return model
```

```python
def predict_and_analyze(model, dataloader, class_names, threshold=None):
    model.eval()
    device = next(model.parameters()).device
    all_preds = []
    all_targets = []

    with torch.no_grad():
        for batch in tqdm(dataloader, desc="Evaluating"):
            images = batch["image"].to(device)
            targets = batch["target"].cpu()

            outputs = model(images)

            if isinstance(outputs, dict):
                logits = outputs.get("logits")
                if logits is None:
                    raise ValueError("Model returned dict without 'logits'")
            else:
                logits = outputs

            probs = torch.sigmoid(logits).cpu()

            if threshold is None:
                thresh = 0.5
            elif isinstance(threshold, (list, np.ndarray)):
                thresh = torch.tensor(threshold, dtype=probs.dtype)
            else:
                thresh = float(threshold)

            if isinstance(thresh, torch.Tensor):
                thresh = thresh.to(probs.device)
                preds = (probs > thresh).int()
            else:
                preds = (probs > thresh).int()

            all_preds.append(preds)
            all_targets.append(targets.int())

    y_pred = torch.cat(all_preds, axis=0).numpy()
    y_true = torch.cat(all_targets, axis=0).numpy()

    print("Classification Report:")
    print(classification_report(y_true, y_pred, target_names=class_names, zero_division=0))

    for i, name in enumerate(class_names):
        cm = confusion_matrix(y_true[:, i], y_pred[:, i])
        print(f"\nConfusion matrix for {name}:\n{cm}")
```

```python
import pandas as pd
from src.cxr_training_pipeline.pipelines.data_science.nodes import (
    make_train_val_indices,
    build_transforms_node,
    prepare_datamodule_config,
    build_datamodule_node,
)

train_val_df = pd.read_csv(r"data/02_intermediate/dataset_splits/train_val.csv")
test_df = pd.read_csv(r"data/02_intermediate/dataset_splits/test.csv")

split_params = {"random_state": 42, "val_size": 0.25}
train_idx, val_idx = make_train_val_indices(train_val_df, split_params)


transform_params = {"random_rotation": 10, "random_translate": (0.02, 0.02), "random_scale": (0.9, 1.1)}
train_tfms, eval_tfms = build_transforms_node(transform_params)

parameters = {
    "kwargs": {
        "train_val_images_path": r"data/02_intermediate/precomputed/224/train_images.npy",
        "test_images_path": r"data/02_intermediate/precomputed/224/test_images.npy",
        "y_train_val_path": r"data/02_intermediate/precomputed/224/train_labels.npy",
        "y_test_path": r"data/02_intermediate/precomputed/224/test_labels.npy",
        "batch_size": 32,
        "num_workers": 0,
    }
}

datamodule_config = prepare_datamodule_config(
    train_val_df=train_val_df,
    test_df=test_df,
    train_idx=train_idx,
    val_idx=val_idx,
    train_tfms=train_tfms,
    eval_tfms=eval_tfms,
    parameters=parameters,
)

datamodule = build_datamodule_node("image_only", datamodule_config)


datamodule.setup(stage="fit")
train_loader = datamodule.train_dataloader()
val_loader = datamodule.val_dataloader()

datamodule.setup(stage="test")
test_loader = datamodule.test_dataloader()

batch = next(iter(val_loader))
print("val batch image shape:", batch["image"].shape, "targets shape:", batch["target"].shape)
print("n_train:", len(train_loader.dataset), "n_val:", len(val_loader.dataset), "n_test:", len(test_loader.dataset))
```
