```python
def _load_tensor_image(tensor_path, device):
    image_tensor = torch.load(tensor_path, map_location="cpu", weights_only=False)
    if isinstance(image_tensor, dict):
        image_tensor = image_tensor.get("image", image_tensor)
    return image_tensor.to(device).unsqueeze(0).float()

def _blend_heatmap(image_array, heatmap, alpha=0.32, cmap="jet"):
    normalized_heatmap = (heatmap - heatmap.min()) / (heatmap.max() - heatmap.min() + 1e-8)
    heatmap_rgba = plt.get_cmap(cmap)(normalized_heatmap)
    heatmap_rgb = heatmap_rgba[..., :3]
    image_float = image_array.astype(np.float32) / 255.0
    blended = (1.0 - alpha) * image_float + alpha * heatmap_rgb
    return np.clip(blended, 0.0, 1.0)


def _image_to_tensor(image, device):
    image_array = np.asarray(image)

    if image_array.ndim == 2:
        # Obraz grayscale: H, W -> 1, H, W
        image_tensor = torch.from_numpy(image_array).unsqueeze(0)

    elif image_array.ndim == 3:
        # Obraz RGB: H, W, C -> C, H, W
        image_tensor = torch.from_numpy(image_array).permute(2, 0, 1)

    else:
        raise ValueError(
            f"Nieobsługiwany kształt obrazu: {image_array.shape}"
        )

    return (
        image_tensor
        .unsqueeze(0)
        .float()
        .div(255.0)
        .to(device)
    )


def _resize_cam_to_image(cam_small, image_size):
    if isinstance(cam_small, np.ndarray):
        cam_small = torch.from_numpy(cam_small)

    if cam_small.ndim == 2:
        cam_small = cam_small.unsqueeze(0)

    return (
        F.interpolate(
            cam_small.unsqueeze(1),
            size=(image_size, image_size),
            mode="bilinear",
            align_corners=False,
        )
        .squeeze(1)
        .squeeze(0)
        .detach()
        .cpu()
        .numpy()
    )


def get_cam_for_class(image, model, class_idx, device):
    image_tensor = _image_to_tensor(image, device)

    model.eval()

    with torch.no_grad():
        image_cam, _ = model.cam(
            image_tensor,
            class_idx=class_idx,
        )

    return image_cam.detach()


def get_cam_for_all_classes(
    image,
    model,
    pathology_list,
    device,
):
    cams = {}

    for class_idx, class_name in enumerate(pathology_list):
        cams[class_name] = get_cam_for_class(
            image=image,
            model=model,
            class_idx=class_idx,
            device=device,
        )

    return cams


def get_grad_cam_for_class(
    image,
    model,
    class_idx,
    device,
):
    image_tensor = _image_to_tensor(image, device)

    model.eval()

    grad_cam, _ = model.grad_cam(
        image=image_tensor,
        class_idx=class_idx,
    )
    return grad_cam.detach()


def get_grad_cam_for_all_classes(
    image,
    model,
    pathology_list,
    device,
):
    grad_cams = {}

    for class_idx, class_name in enumerate(pathology_list):
        grad_cams[class_name] = get_grad_cam_for_class(
            image=image,
            model=model,
            class_idx=class_idx,
            device=device,
        )

    return grad_cams
```

```python
for index in range(len(pathology_list)):
    class_name = pathology_list[index]
    class_idx = index
    image_cam = get_cam_for_class(im, model, class_idx, device)
    image_grad_cam = get_grad_cam_for_class(im, model, class_idx, device)
    heatmap = _resize_cam_to_image(image_cam, 224)
    grad_cam_heatmap = _resize_cam_to_image(image_grad_cam, 224)
    blended_image = _blend_heatmap(np.array(im), heatmap, alpha=0.32, cmap="jet")
    blended_grad_cam_image = _blend_heatmap(np.array(im), grad_cam_heatmap, alpha=0.32, cmap="jet")
    fig, axes = plt.subplots(1, 2, figsize=(12, 6))
    axes[0].imshow(blended_image)
    axes[0].set_title(f"CAM for class: {class_name}")
    axes[0].axis("off")
    axes[1].imshow(blended_grad_cam_image)
    axes[1].set_title(f"Grad-CAM for class: {class_name}")
    axes[1].axis("off")
    plt.show()
```

### HiRes CAM

```python
def normalize_cam(
    cam: torch.Tensor,
    eps: float = 1e-8,
) -> torch.Tensor:
    cam = cam - cam.amin(dim=(1, 2), keepdim=True)
    cam = cam / (
        cam.amax(dim=(1, 2), keepdim=True) + eps
    )
    return cam


def hires_cam_from_output(
    output: dict[str, torch.Tensor],
    class_idx: int,
    retain_graph: bool = False,
) -> torch.Tensor:
    """
    Oblicza HiResCAM na podstawie outputu modelu.

    Oczekiwane elementy output:
        output["logits"]:          [B, num_classes]
        output["transition_maps"]: [B, C, H, W]

    Zwraca:
        Znormalizowaną mapę HiResCAM [B, H, W].
    """
    logits = output["logits"]
    transition_maps = output["transition_maps"]

    if logits.ndim != 2:
        raise ValueError(
            f"logits powinny mieć kształt [B, num_classes], "
            f"otrzymano {tuple(logits.shape)}."
        )

    if transition_maps.ndim != 4:
        raise ValueError(
            f"transition_maps powinny mieć kształt [B, C, H, W], "
            f"otrzymano {tuple(transition_maps.shape)}."
        )

    num_classes = logits.shape[1]

    if not 0 <= class_idx < num_classes:
        raise ValueError(
            f"class_idx musi należeć do zakresu "
            f"[0, {num_classes - 1}], otrzymano {class_idx}."
        )

    if not transition_maps.requires_grad:
        raise RuntimeError(
            "transition_maps nie wymaga gradientów. "
            "Output musi pochodzić z forward passu wykonanego "
            "z włączonym autograd."
        )

    class_score = logits[:, class_idx].sum()

    gradients = torch.autograd.grad(
        outputs=class_score,
        inputs=transition_maps,
        retain_graph=retain_graph,
        create_graph=False,
        allow_unused=False,
    )[0]

    # HiResCAM nie uśrednia gradientów przestrzennie.
    hires_cam = (
        gradients * transition_maps
    ).sum(dim=1)

    hires_cam = F.relu(hires_cam)
    hires_cam = normalize_cam(hires_cam)

    return hires_cam.detach()
```

```python
def _normalize_heatmap(heatmap):
    heatmap = heatmap.astype(np.float32)
    return (heatmap - heatmap.min()) / (heatmap.max() - heatmap.min() + 1e-8)


def get_average_cam_over_classes(image, model, pathology_list, device, image_size=224):
    image_array = np.array(image)
    image_tensor = torch.from_numpy(image_array).permute(2, 0, 1).unsqueeze(0).float().to(device) / 255.0

    class_heatmaps = []
    model.eval()
    with torch.no_grad():
        for class_idx, _ in enumerate(pathology_list):
            cam_small, _ = model.cam(image_tensor, class_idx=class_idx)
            cam = _resize_cam_to_image(cam_small, image_size)
            class_heatmaps.append(_normalize_heatmap(cam))

    average_heatmap = np.mean(np.stack(class_heatmaps, axis=0), axis=0)
    return _normalize_heatmap(average_heatmap)


def plot_average_cam(image, model, pathology_list, device, image_size=224, alpha=0.25, cmap="jet"):
    average_heatmap = get_average_cam_over_classes(
        image=image,
        model=model,
        pathology_list=pathology_list,
        device=device,
        image_size=image_size,
    )
    blended_image = _blend_heatmap(np.array(image), average_heatmap, alpha=alpha, cmap=cmap)

    plt.figure(figsize=(7, 7))
    plt.imshow(blended_image)
    plt.title("Average CAM across all classes")
    plt.axis("off")
    plt.show()

    return average_heatmap

plot_average_cam(im, model, pathology_list, device, image_size=224, alpha=0.25, cmap="jet")
```

```python
def _weighted_average_heatmaps(class_heatmaps, class_weights):
    heatmaps = np.stack(class_heatmaps, axis=0)
    weights = np.asarray(class_weights, dtype=np.float32)

    if weights.sum() <= 1e-8:
        weighted_heatmap = heatmaps.mean(axis=0)
    else:
        weighted_heatmap = np.average(
            heatmaps,
            axis=0,
            weights=weights,
        )

    return _normalize_heatmap(weighted_heatmap)
```

```python
def get_weighted_cam_over_classes(
    image,
    model,
    pathology_list,
    device,
    image_size=224,
):
    image_array = np.asarray(image)

    image_tensor = (
        torch.from_numpy(image_array)
        .permute(2, 0, 1)
        .unsqueeze(0)
        .float()
        .to(device)
        / 255.0
    )

    model.eval()
    class_heatmaps = []

    with torch.no_grad():
        # Prawdopodobieństwa klas używane jako wagi.
        output = model(image_tensor)
        probabilities = (
            torch.sigmoid(output["logits"])[0]
            .detach()
            .cpu()
            .numpy()
        )

        for class_idx in range(len(pathology_list)):
            cam_small, _ = model.cam(
                image_tensor,
                class_idx=class_idx,
            )

            cam = _resize_cam_to_image(
                cam_small,
                image_size,
            )

            class_heatmaps.append(
                _normalize_heatmap(cam)
            )

    return _weighted_average_heatmaps(
        class_heatmaps,
        probabilities,
    )
```

```python
def plot_weighted_cam(
    image,
    model,
    pathology_list,
    device,
    image_size=224,
    alpha=0.25,
    cmap="jet",
):
    weighted_heatmap = get_weighted_cam_over_classes(
        image=image,
        model=model,
        pathology_list=pathology_list,
        device=device,
        image_size=image_size,
    )

    blended_image = _blend_heatmap(
        np.asarray(image),
        weighted_heatmap,
        alpha=alpha,
        cmap=cmap,
    )

    plt.figure(figsize=(7, 7))
    plt.imshow(blended_image)
    plt.title("Probability-weighted CAM")
    plt.axis("off")
    plt.show()

    return weighted_heatmap
```

```python
def get_average_hires_cam_over_classes(
    image,
    model,
    pathology_list,
    device,
    image_size=224,
):
    image_array = np.asarray(image)

    image_tensor = (
        torch.from_numpy(image_array)
        .permute(2, 0, 1)
        .unsqueeze(0)
        .float()
        .div(255.0)
        .to(device)
    )

    model.eval()
    class_heatmaps = []

    with torch.enable_grad():
        output = model(
            image_tensor,
            retain_transition_grad=True,
        )

        for class_idx in range(len(pathology_list)):
            hires_cam = hires_cam_from_output(
                output=output,
                class_idx=class_idx,
                retain_graph=class_idx < len(pathology_list) - 1,
            )

            hires_cam = _resize_cam_to_image(
                hires_cam,
                image_size,
            )

            class_heatmaps.append(hires_cam)

    average_heatmap = np.stack(
        class_heatmaps,
        axis=0,
    ).mean(axis=0)

    return _normalize_heatmap(average_heatmap)


def plot_average_hires_cam(
    image,
    model,
    pathology_list,
    device,
    image_size=224,
    alpha=0.25,
    cmap="jet",
):
    average_heatmap = get_average_hires_cam_over_classes(
        image=image,
        model=model,
        pathology_list=pathology_list,
        device=device,
        image_size=image_size,
    )

    image_array = np.asarray(image)

    # Obraz i heatmapa muszą mieć ten sam rozmiar.
    if image_array.shape[:2] != average_heatmap.shape:
        image_tensor = (
            torch.from_numpy(image_array)
            .permute(2, 0, 1)
            .unsqueeze(0)
            .float()
        )

        image_array = (
            F.interpolate(
                image_tensor,
                size=average_heatmap.shape,
                mode="bilinear",
                align_corners=False,
            )
            .squeeze(0)
            .permute(1, 2, 0)
            .byte()
            .numpy()
        )

    blended_image = _blend_heatmap(
        image_array,
        average_heatmap,
        alpha=alpha,
        cmap=cmap,
    )

    plt.figure(figsize=(7, 7))
    plt.imshow(blended_image)
    plt.title("Average HiResCAM across all classes")
    plt.axis("off")
    plt.show()

    return average_heatmap
```

```python
def get_bboxes_for_image(image_index, bbox_dataframe):

    bbox_row = bbox_dataframe[
        (bbox_dataframe["image_index"] == image_index)
    ]

    if bbox_row.empty:
        raise ValueError(f"Nie znaleziono bounding box dla obrazu {image_index} i etykiety {finding_label}.")

    return bbox_row.iloc[0]

def plot_bboxes_for_image(image_index, bbox_dataframe):
    bbox_row = get_bboxes_for_image(image_index, bbox_dataframe)

    x0, y0, x1, y1 = _scale_bbox_to_image(
        float(bbox_row["x"]), float(bbox_row["y"]),
        float(bbox_row["w"]), float(bbox_row["h"]),
        src=1024, dst=224,
    )

    raw_img = (
        PILImage.open(f"data/01_raw/images/{bbox_row['image_index']}")
        .convert("RGB")
        .resize((224, 224))
    )

    plt.figure(figsize=(7, 7))
    plt.imshow(raw_img)
    plt.gca().add_patch(Rectangle((x0, y0), x1 - x0, y1 - y0,
                                   fill=False, edgecolor="lime", linewidth=2))
    plt.title(f"Bounding box for image: {bbox_row['image_index']}, finding: {bbox_row['finding_label']}")
    plt.axis("off")
    plt.show()

for image_index in image_indexes[:5]:
    plot_bboxes_for_image(image_index, bbox_dataframe)



```

```python
def binarize_cam(cam, mode="top_percent", top_percent=0.20, threshold_value=0.5):
    if mode == "top_percent":
        threshold = np.percentile(cam, 100 * (1 - top_percent))
        return cam >= threshold
    elif mode == "threshold":
        return cam >= threshold_value
    else:
        raise ValueError(f"Nieobsługiwany tryb binarizacji: {mode}")

def compute_iou(pred_mask, gt_mask):
    intersection = np.logical_and(pred_mask, gt_mask).sum()
    union = np.logical_or(pred_mask, gt_mask).sum()
    if union == 0:
        return 1.0 if intersection == 0 else 0.0
    return intersection / union

def get_weighted_cam_over_classes_from_tensor(
    image_tensor,
    model,
    pathology_list,
    image_size=224,
):

    if image_tensor.ndim == 3:
        image_tensor = image_tensor.unsqueeze(0)

    image_tensor = image_tensor.float()

    model.eval()
    class_heatmaps = []

    with torch.no_grad():
        output = model(image_tensor)

        probabilities = (
            torch.sigmoid(output["logits"])[0]
            .detach()
            .cpu()
            .numpy()
        )

        num_classes = min(len(pathology_list), len(probabilities))

        for class_idx in range(num_classes):
            cam_small, _ = model.cam(
                image_tensor,
                class_idx=class_idx,
            )

            cam_full = _resize_cam_to_image(
                cam_small,
                image_size,
            )

            class_heatmaps.append(
                _normalize_heatmap(cam_full)
            )

    weighted_cam = _weighted_average_heatmaps(
        class_heatmaps,
        probabilities[:num_classes],
    )

    return weighted_cam


def load_image_tensor_for_model(
    tensor_path,
    device,
):
    image_tensor = torch.load(
        tensor_path,
        map_location="cpu",
        weights_only=False,
    )

    if isinstance(image_tensor, dict):
        image_tensor = image_tensor.get("image", image_tensor)

    if image_tensor.ndim == 2:
        image_tensor = image_tensor.unsqueeze(0)

    if image_tensor.ndim == 3:
        image_tensor = image_tensor.unsqueeze(0)

    return image_tensor.to(device).float()

def load_image_tensor_for_model(
    tensor_path,
    device,
):
    image_tensor = torch.load(
        tensor_path,
        map_location="cpu",
        weights_only=False,
    )

    if isinstance(image_tensor, dict):
        image_tensor = image_tensor.get("image", image_tensor)

    if image_tensor.ndim == 2:
        image_tensor = image_tensor.unsqueeze(0)

    if image_tensor.ndim == 3:
        image_tensor = image_tensor.unsqueeze(0)

    return image_tensor.to(device).float()

def compute_weighted_cam_iou_for_image(
    image_index,
    tensor_paths,
    bbox_dataframe,
    model,
    pathology_list,
    device,
    localization_params,
    image_size=224,
    bbox_src_size=1024,
):
    class_rows = bbox_dataframe[
        bbox_dataframe["image_index"] == image_index
    ]

    if class_rows.empty:
        return {
            "image_index": image_index,
            "ious": [],
            "mean_iou": np.nan,
            "weighted_cam": None,
            "pred_mask": None,
        }

    tensor_path = tensor_paths[image_index]

    image_tensor = load_image_tensor_for_model(
        tensor_path=tensor_path,
        device=device,
    )

    weighted_cam = get_weighted_cam_over_classes_from_tensor(
        image_tensor=image_tensor,
        model=model,
        pathology_list=pathology_list,
        image_size=image_size,
    )

    pred_mask = binarize_cam(
        weighted_cam,
        mode=localization_params["threshold_mode"],
        top_percent=float(
            localization_params.get("top_percent", 0.20)
        ),
        threshold_value=float(
            localization_params.get("threshold_value", 0.5)
        ),
    )

    ious = []

    for _, bbox_row in class_rows.iterrows():
        x0, y0, x1, y1 = _scale_bbox_to_image(
            float(bbox_row["x"]),
            float(bbox_row["y"]),
            float(bbox_row["w"]),
            float(bbox_row["h"]),
            src=bbox_src_size,
            dst=image_size,
        )

        gt_mask = np.zeros(
            (image_size, image_size),
            dtype=bool,
        )

        gt_mask[y0:y1, x0:x1] = True

        iou = compute_iou(
            pred_mask,
            gt_mask,
        )

        ious.append(
            {
                "finding_label": bbox_row["finding_label"],
                "bbox": [x0, y0, x1, y1],
                "iou": float(iou),
            }
        )

    mean_iou = float(
        np.mean([item["iou"] for item in ious])
    )

    return {
        "image_index": image_index,
        "ious": ious,
        "mean_iou": mean_iou,
        "weighted_cam": weighted_cam,
        "pred_mask": pred_mask,
    }
```

```python
all_results = []

for image_index in image_indexes:
    result = compute_weighted_cam_iou_for_image(
        image_index=image_index,
        tensor_paths=tensor_paths,
        bbox_dataframe=bbox_dataframe,
        model=model,
        pathology_list=pathology_list,
        device=device,
        localization_params=localization_params,
        image_size=224,
        bbox_src_size=1024,
    )

    all_results.append(result)


mean_ious = np.mean([result["mean_iou"] for result in all_results if not np.isnan(result["mean_iou"])])
print(mean_ious)
```
