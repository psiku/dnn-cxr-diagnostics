### Generate iou for image

```python
import numpy as np
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
from PIL import Image as PILImage

from src.cxr_training_pipeline.pipelines.testing.nodes import (
    _binarize_cam,
    _build_model_from_checkpoint,
    _iou_from_masks,
    _resolve_class_index,
    _scale_bbox_to_image,
)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("device:", device)
```

```python
classifier_params = catalog.load("params:training.classifier")
lit_module_params = catalog.load("params:training.lit_module")
pathology_list = catalog.load("params:data_processing.pathology_list")
localization_params = catalog.load("params:localization")
best_checkpoint_path = catalog.load("best_checkpoint_path")

cxr8_bbox = catalog.load("cxr8_bbox")
tensor_test_dataset = catalog.load("tensor_test_dataset")
tensor_paths = tensor_test_dataset.set_index("image_index")["image_tensor_path"].to_dict()
```

```python
model = _build_model_from_checkpoint(
    classifier_params=classifier_params,
    lit_module_params=lit_module_params,
    checkpoint_path=best_checkpoint_path,
    device=device,
)
print(type(model).__name__, "ready on", device)
```

```python
# Pick which (image, finding) to visualize.
# Default: highest-IoU Cardiomegaly example, since that class scores best.
TARGET_CLASS = "Cardiomegaly"
RANK = -1  # 0 = best IoU; increase to see weaker examples

target = (
    localization_iou_per_image
    .query("class_name == @TARGET_CLASS and status == 'ok'")
    .sort_values("iou", ascending=False)
    .iloc[RANK]
)

bbox_row = cxr8_bbox[
    (cxr8_bbox["image_index"] == target["image_index"])
    & (cxr8_bbox["finding_label"] == target["finding_label"])
].iloc[0]

print(f"image: {bbox_row['image_index']}  finding: {bbox_row['finding_label']}")
print(f"reported IoU from pipeline: {target['iou']:.4f}")
bbox_row
```

```python
image_size = int(localization_params["image_size"])
original_size = int(localization_params["original_size"])

class_idx = _resolve_class_index(bbox_row["finding_label"], pathology_list)
assert class_idx is not None, f"{bbox_row['finding_label']} not in pathology_list"

# Same tensor-loading path the pipeline uses, so the CAM matches the parquet IoU.
tensor_path = tensor_paths[bbox_row["image_index"]]
image_tensor = torch.load(tensor_path, map_location="cpu", weights_only=False)
if isinstance(image_tensor, dict):
    image_tensor = image_tensor.get("image", image_tensor)
image_tensor = image_tensor.to(device).unsqueeze(0).float()

cam_small, _ = model.cam(image_tensor, class_idx=class_idx)
cam_full = (
    F.interpolate(
        cam_small.unsqueeze(1),
        size=(image_size, image_size),
        mode="bilinear",
        align_corners=False,
    )
    .squeeze(1)
    .squeeze(0)
    .detach()
    .cpu()
    .numpy()
)

pred_mask = _binarize_cam(
    cam_full,
    mode=localization_params["threshold_mode"],
    top_percent=float(localization_params.get("top_percent", 0.20)),
    value=float(localization_params.get("threshold_value", 0.5)),
)

x0, y0, x1, y1 = _scale_bbox_to_image(
    float(bbox_row["x"]), float(bbox_row["y"]),
    float(bbox_row["w"]), float(bbox_row["h"]),
    src=original_size, dst=image_size,
)
gt_mask = np.zeros((image_size, image_size), dtype=bool)
gt_mask[y0:y1, x0:x1] = True

iou = _iou_from_masks(pred_mask, gt_mask)
print(f"computed IoU: {iou:.4f}  (pipeline-reported: {target['iou']:.4f})")
print(f"CAM shape (backbone): {tuple(cam_small.shape)}  ->  upsampled to {cam_full.shape}")
```

```python
raw_img = (
    PILImage.open(f"data/01_raw/images/{bbox_row['image_index']}")
    .convert("RGB")
    .resize((image_size, image_size))
)
raw_arr = np.array(raw_img)

# Color-code the IoU breakdown: green = TP, red = FP (CAM only), blue = FN (bbox only).
breakdown = np.zeros((*pred_mask.shape, 3), dtype=np.uint8)
breakdown[pred_mask & ~gt_mask] = (255, 0, 0)
breakdown[~pred_mask & gt_mask] = (0, 0, 255)
breakdown[pred_mask & gt_mask] = (0, 255, 0)

fig, axes = plt.subplots(1, 4, figsize=(20, 5))

axes[0].imshow(raw_arr)
axes[0].add_patch(Rectangle((x0, y0), x1 - x0, y1 - y0,
                             fill=False, edgecolor="lime", linewidth=2))
axes[0].set_title(f"{bbox_row['image_index']}\n{bbox_row['finding_label']}")
axes[0].axis("off")

axes[1].imshow(raw_arr)
axes[1].imshow(cam_full, cmap="jet", alpha=0.5)
axes[1].add_patch(Rectangle((x0, y0), x1 - x0, y1 - y0,
                             fill=False, edgecolor="lime", linewidth=2))
axes[1].set_title(f"CAM overlay (class={pathology_list[class_idx]})")
axes[1].axis("off")

axes[2].imshow(pred_mask, cmap="gray")
axes[2].add_patch(Rectangle((x0, y0), x1 - x0, y1 - y0,
                             fill=False, edgecolor="lime", linewidth=2))
top_pct = int(float(localization_params.get("top_percent", 0.20)) * 100)
axes[2].set_title(f"Thresholded CAM (top {top_pct}%)")
axes[2].axis("off")

axes[3].imshow(raw_arr)
axes[3].imshow(breakdown, alpha=0.45)
axes[3].set_title(f"IoU = {iou:.4f}  (green=TP, red=FP, blue=FN)")
axes[3].axis("off")

fig.tight_layout()
plt.show()
```
