```python
def create_f1_plot(
    thresholds_search_results: pd.DataFrame,
    pathology: str,
    threshold_list,
    ax: plt.Axes,
) -> None:
    """Draw F1/recall/precision for one pathology on a provided axis."""
    subset = thresholds_search_results[thresholds_search_results["class_name"] == pathology]
    best_threshold = subset.loc[subset["f1"].idxmax(), "threshold"]

    ax.plot(subset["threshold"], subset["f1"], label="F1 Score")
    ax.plot(subset["threshold"], subset["recall"], label="Recall")
    ax.plot(subset["threshold"], subset["precision"], label="Precision")

    ax.set_title(pathology)
    ax.set_xlabel("Threshold")
    ax.set_ylabel("Score")
    ax.set_xticks(threshold_list[::5])
    ax.axvline(best_threshold, color="red", linestyle="--", label="Best Threshold")
    ax.tick_params(axis="x", rotation=45)
    ax.grid(True)
```

```python
fig, axes = plt.subplots(3, 4, figsize=(45, 30))
for ax, pathology in zip(axes.flatten(), pathologies):
    create_f1_plot(thresholds_search_results, pathology, threshold_list, ax)
```
