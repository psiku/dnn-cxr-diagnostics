```python
from matplotlib.ticker import MaxNLocator


def create_f1_plot(
    thresholds_search_results: pd.DataFrame,
    pathology: str,
    ax: plt.Axes,
) -> None:
    """Draw F1/recall/precision for one pathology on a provided axis."""
    subset = (
        thresholds_search_results[thresholds_search_results["class_name"] == pathology]
        .sort_values("threshold")
    )

    # Prefer the threshold actually picked by the tuner; fall back to argmax F1.
    if "chosen" in subset.columns and subset["chosen"].any():
        best_threshold = float(subset.loc[subset["chosen"], "threshold"].iloc[0])
        chosen_label = "Chosen threshold"
    else:
        best_threshold = float(subset.loc[subset["f1"].idxmax(), "threshold"])
        chosen_label = "Best threshold based on F1 Score"

    ax.plot(subset["threshold"], subset["f1"], label="F1 Score")
    ax.plot(subset["threshold"], subset["recall"], label="Recall")
    ax.plot(subset["threshold"], subset["precision"], label="Precision")

    ax.set_title(pathology)
    ax.set_xlabel("Threshold")
    ax.set_ylabel("Score")
    ax.set_xlim(0.0, 1.0)
    ax.set_ylim(0.0, 1.0)
    ax.xaxis.set_major_locator(MaxNLocator(nbins=10))
    ax.axvline(best_threshold, color="red", linestyle="--", label=chosen_label)
    ax.tick_params(axis="x", rotation=45)
    ax.grid(True)
    ax.legend()
```

```python
fig, axes = plt.subplots(3, 4, figsize=(45, 30))
for ax, pathology in zip(axes.flatten(), pathologies):
    create_f1_plot(thresholds_search_results, pathology, ax)
```
