### Mapping

```python
class SexMapping:
    MAPPING = {"M": 1, "F": 0}

class ViewPositionMapping:
    MAPPING = {"AP": 1, "PA": 0}

def encode_sex(x: str) -> float:
    return SexMapping.MAPPING.get(x, np.nan)

def encode_view_position(x: str) -> float:
    return ViewPositionMapping.MAPPING.get(x, np.nan)
```

### Precompute arrays

```python
PATHOLOGY_LIST = ["Infiltration", "Effusion", "Atelectasis", "Nodule", "Mass", "Pneumothorax", "Consolidation", "Pleural_Thickening", "Cardiomegaly", "Emphysema", "Edema", "Fibrosis", "Pneumonia", "Hernia"]

def precompute_arrays(df: pd.DataFrame) -> tuple[np.ndarray, np.ndarray]:
    # metadata
    age = df['patient_age'].fillna(-1).values.astype(np.float32)
    age = np.clip(age, 0, 120) / 120.0 # age normalization

    # gender
    gender = df['patient_sex'].fillna("U").map(encode_sex).astype(np.float32)

    # view position
    view_position = df['view_position'].fillna("U").map(encode_view_position).astype(np.float32)

    x_meta = np.stack([age, gender, view_position], axis=1).astype(np.float32)
    y_labels = df[PATHOLOGY_LIST].values.astype(np.float32)

    return x_meta, y_labels
```

```python
out_dir = Path("data/02_intermediate/precomputed_metadata")
out_dir.mkdir(parents=True, exist_ok=True)
```

```python
x_train_val, y_train_val = precompute_arrays(train_val_df)
x_test, y_test = precompute_arrays(test_df)

np.save(out_dir / "x_train_val.npy", x_train_val)
np.save(out_dir / "y_train_val.npy", y_train_val)

np.save(out_dir / "x_test.npy", x_test)
np.save(out_dir / "y_test.npy", y_test)
```
