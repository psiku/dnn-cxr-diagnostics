### Dataframe cleaning

This metadata contains a significant amount of useful information; however, not all of it will be used by our classifier.

1. Columns that will be used:

image_index – identifier of the image, used to locate the corresponding file in the dataset

finding_labels – labels describing the pathologies present in a given image

view_position – indicates whether the image was taken in the PA or AP view, which can influence the appearance of certain structures in chest X-rays

patient_sex – may provide additional context, as some conditions have different prevalence depending on sex

patient_age – relevant because the probability of certain pathologies increases with age

2. Columns that will not be used:

follow_up – may indirectly indicate that a patient has already been examined multiple times, which could introduce bias

patient_id – could lead to data leakage if the same patient appears in both training and test sets

original image resolution – images will be resized during preprocessing, so the original resolution does not provide useful information for the model
