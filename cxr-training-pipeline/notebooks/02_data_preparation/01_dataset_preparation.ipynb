### Dataframe cleaning

This metadata contains a significant amount of useful information; however, not all of it will be used by our classifier.

1. Columns that will be used:

- image_index – identifier of the image, used to locate the corresponding file in the dataset

- finding_labels – labels describing the pathologies present in a given image

- view_position – indicates whether the image was taken in the PA or AP view, which can influence the appearance of certain structures in chest X-rays

- patient_sex – may provide additional context, as some conditions have different prevalence depending on sex

- patient_age – relevant because the probability of certain pathologies increases with age

2. Columns that will not be used:

- follow_up – may indirectly indicate that a patient has already been examined multiple times, which could introduce bias

- patient_id – could lead to data leakage if the same patient appears in both training and test sets

- original image resolution – images will be resized during preprocessing, so the original resolution does not provide useful information for the model

```python
cleaned_metadata = metadata[["image_index", "finding_labels", "patient_age", "patient_sex", "view_position"]]
```

```python
cleaned_metadata.head()
```

### Encoding finding labels

Since we are dealing with a multi-label classification problem, the finding_labels column must be split into separate labels and converted into a vector representation. Each image will be associated with a pathology vector, where every position corresponds to a specific pathology.

A value of 1 in a given position indicates that the pathology is present in the image, while a value of 0 indicates that it is not present.

```python
PATHOLOGY_LIST = [
    "Infiltration",
    "Effusion",
    "Atelectasis",
    "Nodule",
    "Mass",
    "Pneumothorax",
    "Consolidation",
    "Pleural_Thickening",
    "Cardiomegaly",
    "Emphysema",
    "Edema",
    "Fibrosis",
    "Pneumonia",
    "Hernia",
]
```

```python
def labels_to_vector(label_string):
    labels = label_string.split("|")
    return [1 if pathology in labels else 0 for pathology in PATHOLOGY_LIST]


cleaned_metadata["pathology_vector"] = cleaned_metadata["finding_labels"].apply(labels_to_vector)

cleaned_metadata.head()
```

### OHE

```python
labels = cleaned_metadata["finding_labels"].str.get_dummies(sep="|")

cleaned_metadata = cleaned_metadata.join(labels[PATHOLOGY_LIST])

cleaned_metadata.head()
```
