# Tensor Full

```python
def convert_to_tensor(image_path: str, transformation: transforms.Compose) -> torch.Tensor:
    image = Image.open(image_path).convert("RGB")
    tensor_image = transformation(image)
    return tensor_image
```

```python
transformation = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
])

for image_index in tqdm(xray_train_val_df["image_index"], desc="Processing training and validation images"):
    src = os.path.join("data/01_raw/images", image_index)
    image_tensor = convert_to_tensor(src, transformation)
    dst = os.path.join(tensor_full_224, image_index.replace(".png", ".pt"))
    torch.save(image_tensor, dst)

for image_index in tqdm(xray_test_df["image_index"], desc="Processing test images"):
    src = os.path.join("data/01_raw/images", image_index)
    image_tensor = convert_to_tensor(src, transformation)
    dst = os.path.join(tensor_full_224, image_index.replace(".png", ".pt"))
    torch.save(image_tensor, dst)
```

# Tensor Cropped + Mask

```python
def create_cropped_and_mask_tensors(image_path: str, image_index: str, masks_df: pd.DataFrame, out_size: int = 224):
    img_pil = Image.open(image_path).convert("RGB")
    img_np = np.array(img_pil)
    height, width = img_np.shape[:2]

    combined_mask = get_combined_mask_by_image_index(
        image_index=image_index,
        segmentation_masks_df=masks_df,
        height=height,
        width=width
    )

    bbox = get_bbox_coordinates(combined_mask)

    if bbox is not None:
        bbox = expand_bbox(bbox, img_shape=(height, width), margin_ratio=0.01)
        x_min, y_min, x_max, y_max = bbox

        cropped_img_np = img_np[y_min:y_max, x_min:x_max]
        cropped_mask_np = combined_mask[y_min:y_max, x_min:x_max]
    else:
        cropped_img_np = img_np
        cropped_mask_np = combined_mask

    to_tensor = v2.Compose([
        v2.ToImage(),
        v2.ToDtype(torch.float32, scale=True)
    ])


    img_resized = Image.fromarray(cropped_img_np).resize((out_size, out_size), Image.BILINEAR)
    img_tensor = to_tensor(img_resized) # Shape: [3, 224, 224]

    mask_resized = Image.fromarray(cropped_mask_np * 255).resize((out_size, out_size), Image.NEAREST)
    mask_tensor = to_tensor(mask_resized) # Shape: [1, 224, 224]

    return img_tensor, mask_tensor
```

```python
img, mask = create_cropped_and_mask_tensors(
    image_path=os.path.join("data/01_raw/images", xray_train_val_df.iloc[0]["image_index"]),
    image_index=xray_train_val_df.iloc[0]["image_index"],
    masks_df=masks_df,
    out_size=224
)
```

```python
for image_index in tqdm(xray_train_val_df["image_index"], desc="Processing training and validation images for cropping and masking"):
    src = os.path.join("data/01_raw/images", image_index)
    img_tensor, mask_tensor = create_cropped_and_mask_tensors(src, image_index, masks_df)

    dst_img = os.path.join(tensor_cropped_224, image_index.replace(".png", "_cropped.pt"))
    dst_mask = os.path.join(masks_cropped_224, image_index.replace(".png", "_mask.pt"))

    torch.save(img_tensor, dst_img)
    torch.save(mask_tensor, dst_mask)
```
