# Dataset Class

```python
class TensorCXRDataset(Dataset):
    def __init__(
        self,
        df: pd.DataFrame,
        path_col: str,
        label_cols: list[str],
        mask_path_col: str = None,
        transform=None,
    ):
        self.df = df.reset_index(drop=True)
        self.path_col = path_col
        self.mask_path_col = mask_path_col
        self.label_cols = label_cols
        self.transform = transform

        assert self.path_col in self.df.columns, f"Path column '{self.path_col}' missing."
        if self.mask_path_col:
            assert self.mask_path_col in self.df.columns, f"Mask Path column '{self.mask_path_col}' missing."

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx: int):
        row = self.df.iloc[idx]
        image_index = row["image_index"]

        # Load Image
        tensor_path = row[self.path_col]
        img_tensor = torch.load(tensor_path, weights_only=False) # Shape: [C, H, W]

        # Load Mask and combine
        if self.mask_path_col and pd.notna(row[self.mask_path_col]):
            mask_path = row[self.mask_path_col]
            mask_tensor = torch.load(mask_path, weights_only=False) # Assumed shape: [1, H, W]

            # Stack along channel dimension -> Shape: [C+1, H, W]
            img_tensor = torch.cat([img_tensor, mask_tensor], dim=0)

        if self.transform is not None:
            img_tensor = self.transform(img_tensor)

        target = torch.tensor(row[self.label_cols].values.astype(np.float32), dtype=torch.float32)

        return {
            "image": img_tensor,
            "target": target,
            "image_index": image_index,
        }
```

### Dataset test with only images

```python
train_val_dataset = TensorCXRDataset(
    df=combined_train_val_df,
    path_col="image_tensor_path",
    mask_path_col=None,
    label_cols=pathologies,
    transform=None
)

train_val_dataset_cropped = TensorCXRDataset(
    df=combined_train_val_df,
    path_col="cropped_image_tensor_path",
    mask_path_col=None,
    label_cols=pathologies,
    transform=None
)

train_val_dataset_with_masks = TensorCXRDataset(
    df=combined_train_val_df,
    path_col="cropped_image_tensor_path",
    mask_path_col="mask_tensor_path",
    label_cols=pathologies,
    transform=None
)
```

```python
def visualize_dataset_example(dataset: TensorCXRDataset, idx: int):
    sample = dataset[idx]
    image_tensor = sample["image"]
    target = sample["target"]
    image_index = sample["image_index"]

    print(f"Image Index: {image_index}")
    print(f"Target Labels: {target}")

    if dataset.mask_path_col:
        img_channels = image_tensor.shape[0] - 1
        img_tensor = image_tensor[:img_channels, :, :]
        mask_tensor = image_tensor[img_channels:, :, :]

        img_np = img_tensor.permute(1, 2, 0).numpy()  # [H, W, C]
        mask_np = mask_tensor.squeeze(0).numpy()      # [H, W]

        plt.figure(figsize=(10, 5))
        plt.subplot(1, 2, 1)
        plt.imshow(img_np)
        plt.title("Image with Mask Channel")
        plt.axis('off')

        plt.subplot(1, 2, 2)
        plt.imshow(mask_np, cmap='gray')
        plt.title("Segmentation Mask")
        plt.axis('off')

        plt.show()
    else:
        img_np = image_tensor.permute(1, 2, 0).numpy()  # [H, W, C]

        plt.figure(figsize=(5, 5))
        plt.imshow(img_np)
        plt.title("Image")
        plt.axis('off')
        plt.show()
```

```python
visualize_dataset_example(train_val_dataset_cropped, idx=0)
visualize_dataset_example(train_val_dataset, idx=0)
visualize_dataset_example(train_val_dataset_with_masks, idx=0)
```
