```python
import yaml
import pandas as pd
from pathlib import Path
from cxr_training_pipeline.pipelines.data_science.nodes import make_train_val_indices

df = pd.read_csv("data/05_model_input/tensor_train_val.csv")

params = yaml.safe_load(Path("conf/base/parameters_data_science.yml").read_text())["training"]["split"]

baseline_params = dict(params)
baseline_params["class_rebalance"] = dict(params.get("class_rebalance", {}))
baseline_params["class_rebalance"]["enabled"] = False

train_base, val_base = make_train_val_indices(df, baseline_params)

train_reb, val_reb = make_train_val_indices(df, params)

def count_stats(dataframe, idx):
    sub = dataframe.iloc[idx]
    no_finding = (sub["finding_labels"].fillna("").str.strip() == "No Finding").sum()
    hernia = sub["Hernia"].sum() if "Hernia" in sub.columns else None
    pneumonia = sub["Pneumonia"].sum() if "Pneumonia" in sub.columns else None
    return {
        "n_samples": len(sub),
        "no_finding": int(no_finding),
        "Hernia+": int(hernia) if hernia is not None else None,
        "Pneumonia+": int(pneumonia) if pneumonia is not None else None,
    }

print("BASE:", count_stats(df, train_base))
print("REBALANCED:", count_stats(df, train_reb))
print("VAL same size =", len(val_base) == len(val_reb), len(val_reb))
```
