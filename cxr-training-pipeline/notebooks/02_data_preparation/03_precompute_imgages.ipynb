### Mapping

```python
def precompute_images_to_npy(
    df: pd.DataFrame,
    images_dir: str,
    output_path: str,
    image_size: int = 512,
    image_col: str = "image_index",
):
    images_dir = Path(images_dir)
    output_path = Path(output_path)

    n = len(df)

    # tworzysz jedną dużą tablicę [N, H, W]
    arr = np.memmap(
        output_path,
        dtype="uint8",
        mode="w+",
        shape=(n, image_size, image_size),
    )

    for i, image_name in enumerate(tqdm(df[image_col].tolist(), total=n)):
        img_path = images_dir / image_name

        img = Image.open(img_path).convert("L")  # grayscale
        img = img.resize((image_size, image_size))
        img = np.array(img, dtype=np.uint8)

        arr[i] = img

    arr.flush()
    print(f"Saved memmap to: {output_path}")
```

### Precompute images

```python
precompute_images_to_npy(
    df=train_val_df,
    images_dir="data/01_raw/images",
    output_path="data/02_intermediate/precomputed_images/train_val_images.npy",
    image_size=512,
    image_col="image_index",
)
```
