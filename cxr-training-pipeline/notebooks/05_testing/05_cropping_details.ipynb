```python
masks_df = catalog.load("cxr8_segmentation_masks")
images_dir = Path("data/01_raw/images")

# Same margin as data_processing._get_cropped_arrays
MARGIN_RATIO = 0.01

print(f"Masks: {len(masks_df):,} rows")
print(f"Images dir: {images_dir.resolve()} (exists={images_dir.exists()})")
print(f"Unique sizes: {masks_df[['Height', 'Width']].drop_duplicates().to_dict('records')}")
masks_df.head(3)
```

```python
from cxr_training_pipeline.utils.segmentation_utils import rle_to_mask


def crop_area_stats_from_row(row: pd.Series, margin_ratio: float = MARGIN_RATIO) -> dict:
    """Original vs mask-bbox crop area for one segmentation row."""
    image_index = row["Image Index"]
    height = int(row["Height"])
    width = int(row["Width"])
    original_area = height * width

    organ_masks = [
        rle_to_mask(row.get(col, ""), height, width)
        for col in ("Left Lung", "Right Lung", "Heart")
    ]
    combined_mask = np.maximum.reduce(organ_masks)
    bbox = get_bbox_coordinates(combined_mask)

    if bbox is None:
        cropped_area = original_area
        status = "empty_mask"
        x_min = y_min = x_max = y_max = None
    else:
        bbox = expand_bbox(bbox, img_shape=(height, width), margin_ratio=margin_ratio)
        x_min, y_min, x_max, y_max = bbox
        cropped_area = (x_max - x_min) * (y_max - y_min)
        status = "ok"

    fraction_kept = cropped_area / original_area if original_area > 0 else float("nan")
    fraction_saved = 1.0 - fraction_kept

    return {
        "image_index": image_index,
        "height": height,
        "width": width,
        "original_area": original_area,
        "cropped_area": cropped_area,
        "fraction_kept": fraction_kept,
        "fraction_saved": fraction_saved,
        "bbox_w": None if x_min is None else (x_max - x_min),
        "bbox_h": None if y_min is None else (y_max - y_min),
        "status": status,
    }


rows = [
    crop_area_stats_from_row(row)
    for _, row in tqdm(masks_df.iterrows(), total=len(masks_df), desc="Crop area stats")
]
area_df = pd.DataFrame(rows)
area_df.head()
```

```python
scored = area_df[area_df["status"] == "ok"]

summary = {
    "n_images": int(len(area_df)),
    "n_with_mask_bbox": int(len(scored)),
    "n_empty_mask": int((area_df["status"] == "empty_mask").sum()),
    "mean_fraction_kept": float(scored["fraction_kept"].mean()),
    "median_fraction_kept": float(scored["fraction_kept"].median()),
    "mean_fraction_saved": float(scored["fraction_saved"].mean()),
    "median_fraction_saved": float(scored["fraction_saved"].median()),
    "mean_cropped_area_px": float(scored["cropped_area"].mean()),
    "mean_original_area_px": float(scored["original_area"].mean()),
}

print("Mask-based crop area savings (margin_ratio=0.01, same as pipeline)")
print("-" * 60)
for k, v in summary.items():
    if "fraction" in k:
        print(f"{k}: {v:.4f} ({100 * v:.2f}%)")
    else:
        print(f"{k}: {v:,.0f}" if isinstance(v, float) and v > 100 else f"{k}: {v}")

print(
    f"\nOn average, cropping to the mask bbox keeps "
    f"{100 * summary['mean_fraction_kept']:.1f}% of the image "
    f"and removes {100 * summary['mean_fraction_saved']:.1f}% of the pixels."
)

scored[["fraction_kept", "fraction_saved", "bbox_w", "bbox_h"]].describe()
```
