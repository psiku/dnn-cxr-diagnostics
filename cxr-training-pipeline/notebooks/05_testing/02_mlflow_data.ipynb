```python
import matplotlib.pyplot as plt

def get_train_loss_history(run_id: str, mlflow_client: mlflow.tracking.MlflowClient, metric_name) -> None:
    metric_history = mlflow_client.get_metric_history(run_id, metric_name)
    metric_history_sorted = sorted(metric_history, key=lambda metric: metric.step)

    steps = [metric.step for metric in metric_history_sorted]
    values = [metric.value for metric in metric_history_sorted]

    return steps, values

def plot_train_loss_history_for_all_runs(mlflow_client: mlflow.tracking.MlflowClient) -> None:

    plt.figure(figsize=(10, 5))
    for model, run_id in MAPPINGS.items():
        metric_name = "val/loss" if model != "DeiT" else "val_loss"
        steps, values = get_train_loss_history(run_id, mlflow_client, metric_name)

        plt.plot(steps, values, label=model)

    plt.title(f'Training Loss History for Models without scaling')
    plt.xlabel('Step')
    plt.ylabel('Loss')
    plt.grid()
    plt.legend()
    plt.show()

plot_train_loss_history_for_all_runs(mlflow_client)
```

```python
import matplotlib.pyplot as plt
import math

def get_train_loss_history(run_id: str, mlflow_client: mlflow.tracking.MlflowClient, metric_name) -> None:
    metric_history = mlflow_client.get_metric_history(run_id, metric_name)
    metric_history_sorted = sorted(metric_history, key=lambda metric: metric.step)
    steps = []
    values = []

    for metric in metric_history_sorted:
        # 2. Warunek bezpiecznego odrzucenia wartości None i NaN
        if metric.value is not None and not math.isnan(metric.value):
            steps.append(metric.step)
            values.append(metric.value)

    return steps, values
    return steps, values

def plot_loss_history_combined(mlflow_client: mlflow.tracking.MlflowClient, mapping: dict, line_x: float) -> None:
    # 1. Ustawienie profesjonalnego stylu i czcionek (idealne do druku/szarej siatki)
    plt.style.use('seaborn-v0_8-whitegrid')
    plt.rcParams.update({
        'font.size': 11,
        'font.family': 'serif',
        'axes.labelsize': 12,
        'axes.titlesize': 13
    })

    # 2. Tworzenie dwóch wykresów obok siebie z tą samą skalą Y (sharey=True)
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18,8), sharey=True)

    # Paleta barw – upewnij się, że modele mają przypisane stałe kolory na obu wykresach
    colors = plt.cm.tab20(range(len(mapping)))

    for idx, (model, run_id) in enumerate(mapping.items()):
        color = colors[idx]

        # --- WYKRES 1: TRAIN LOSS ---
        train_metric = "train/loss" if model != "DeiT" else "train_loss"
        t_steps, t_values = get_train_loss_history(run_id, mlflow_client, train_metric)

        # Rysujemy linię (z lekko wygładzonym antyaliasingem)
        ax1.plot(t_steps, t_values, label=model, color=color, linewidth=2)

        # --- WYKRES 2: VAL LOSS ---
        val_metric = "val/loss" if model != "DeiT" else "val_loss"
        v_steps, v_values = get_train_loss_history(run_id, mlflow_client, val_metric)

        ax2.plot(v_steps, v_values, label=model, color=color, linewidth=2)

        if v_steps and v_values:
            ax2.scatter(v_steps[-1], v_values[-1], color=color, marker='o', s=40, zorder=5)

        if t_steps and t_values:
            ax1.scatter(t_steps[-1], t_values[-1], color=color, marker='o', s=40, zorder=5)


    # 3. Formatowanie lewego wykresu (Train)
    ax1.set_title('A: Strata zbioru treningowego (ang. Training Loss)', pad=15, fontsize=13)
    ax1.set_xlabel('Krok (ang. Step)', fontsize=12)
    ax1.set_ylabel('Wartość funkcji straty (ang. Loss)', fontsize=12)
    ax1.grid(True, linestyle='--', alpha=0.7)
    ax1.set_xlim(0, 60)
    ax1.set_ylim(0, 2.2)

    # 4. Formatowanie prawego wykresu (Val) + Legenda
    ax2.set_title('B: Strata zbioru walidacyjnego (ang. Validation Loss)', pad=15, fontsize=13)
    ax2.set_xlabel('Krok (ang. Step)', fontsize=12)
    ax2.set_ylabel('Wartość funkcji straty (ang. Loss)', fontsize=12)
    ax2.grid(True, linestyle='--', alpha=0.7)
    ax2.set_xlim(0, 60)
    ax2.set_ylim(0, 2.2)
    # Umieszczenie legendy pomiędzy wykresami lub po prawej stronie, żeby nie zasłaniała linii
    ax2.legend(
    loc='upper right',
    frameon=True,
    shadow=True
)
    ax2.axvline(x=line_x, color='darkred', linestyle='--', linewidth=1.5, label='Początek przeuczenia')
    ax2.yaxis.set_tick_params(labelleft=True)

    # Dodanie napisu nad/obok linii, żeby wykres sam się tłumaczył
    ax2.text(
    x=10.5,                                # Lekko przesunięte w prawo od linii x=8
    y=ax2.get_ylim()[1] * 1,           # 100% wysokości osi Y (blisko samej górnej krawędzi)
    s='Strefa przeuczenia (ang. Overfitting)',
    color='darkred',
    fontsize=10,
    weight='bold',
    rotation=0,                         # Obrót o 360 stopni sprawia, że czytamy od góry do dołu
    va='top',                             # Kotwica tekstu na jego "górze" (czyli zaczyna się od punktu Y)
    ha='left'                             # Wyrównanie do lewej strony punktu X
)

    # Tytuł główny całej grafiki
    plt.suptitle('Porównanie historii uczenia i efektu przeuczenia modeli korzystających z przycinania zdjęć',
                 fontsize=16, weight='bold', y=0.98)

    # 5. Zapis do pliku wektorowego PDF (w pracy inżynierskiej nie straci jakości)
    plt.tight_layout()
    plt.savefig('historia_strat_overfitting.pdf', bbox_inches='tight', dpi=300)
    plt.show()
```

### AUROC i AP

```python
import matplotlib.pyplot as plt
import math
import mlflow


def get_metric_history_values(
    run_id: str,
    mlflow_client: mlflow.tracking.MlflowClient,
    metric_name: str
):
    metric_history = mlflow_client.get_metric_history(run_id, metric_name)
    metric_history_sorted = sorted(metric_history, key=lambda metric: metric.step)

    steps = []
    values = []

    for metric in metric_history_sorted:
        if metric.value is not None and math.isfinite(float(metric.value)):
            steps.append(metric.step)
            values.append(metric.value)

    return steps, values


def plot_validation_metrics_combined(
    mlflow_client: mlflow.tracking.MlflowClient,
    mapping: dict,
    line_x: float | None = None,
    title: str = "Tytuł"
) -> None:
    plt.style.use('seaborn-v0_8-whitegrid')
    plt.rcParams.update({
        'font.size': 11,
        'font.family': 'serif',
        'axes.labelsize': 12,
        'axes.titlesize': 13
    })

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 8), sharey=False)

    colors = plt.cm.tab20(range(len(mapping)))

    for idx, (model, run_id) in enumerate(mapping.items()):
        color = colors[idx]

        # --- WYKRES 1: VAL AUROC MACRO ---
        auroc_metric = "val_auroc" if model == "DeiT" else "val/auroc_macro"
        auroc_steps, auroc_values = get_metric_history_values(
            run_id,
            mlflow_client,
            auroc_metric
        )

        ax1.plot(
            auroc_steps,
            auroc_values,
            label=model,
            color=color,
            linewidth=2
        )

        if auroc_steps and auroc_values:
            ax1.scatter(
                auroc_steps[-1],
                auroc_values[-1],
                color=color,
                marker='o',
                s=40,
                zorder=5
            )

        # --- WYKRES 2: VAL AP MACRO ---
        ap_metric = "val_ap" if model == "DeiT" else "val/ap_macro"
        ap_steps, ap_values = get_metric_history_values(
            run_id,
            mlflow_client,
            ap_metric
        )

        ax2.plot(
            ap_steps,
            ap_values,
            label=model,
            color=color,
            linewidth=2
        )

        if ap_steps and ap_values:
            ax2.scatter(
                ap_steps[-1],
                ap_values[-1],
                color=color,
                marker='o',
                s=40,
                zorder=5
            )

    # --- Formatowanie wykresu AUROC ---
    ax1.set_title('A: AUROC macro na zbiorze walidacyjnym', pad=15, fontsize=13)
    ax1.set_xlabel('Krok (ang. Step)', fontsize=12)
    ax1.set_ylabel('Wartość metryki', fontsize=12)
    ax1.grid(True, linestyle='--', alpha=0.7)
    ax1.set_xlim(0, 60)

    # --- Formatowanie wykresu AP ---
    ax2.set_title('B: AP macro na zbiorze walidacyjnym', pad=15, fontsize=13)
    ax2.set_xlabel('Krok (ang. Step)', fontsize=12)
    ax2.set_ylabel('Wartość metryki', fontsize=12)
    ax2.grid(True, linestyle='--', alpha=0.7)
    ax2.set_xlim(0, 60)
    ax2.yaxis.set_tick_params(labelleft=True)

    # Opcjonalna linia pionowa, np. początek przeuczenia
    if line_x is not None:
        ax1.axvline(
            x=line_x,
            color='darkred',
            linestyle='--',
            linewidth=1.5,
            label='Początek przeuczenia'
        )

        ax2.axvline(
            x=line_x,
            color='darkred',
            linestyle='--',
            linewidth=1.5,
            label='Początek przeuczenia'
        )

        ax2.text(
            x=line_x + 2.5,
            y=ax2.get_ylim()[1],
            s='Strefa przeuczenia',
            color='darkred',
            fontsize=10,
            weight='bold',
            va='top',
            ha='left',
            alpha=0.7
        )

    ax2.legend(
    loc='lower right',
    frameon=True,
    shadow=True
)

    plt.suptitle(
        title,
        fontsize=16,
        weight='bold',
        y=0.98
    )

    plt.tight_layout()
    plt.savefig(
        'historia_metryk_walidacyjnych_auroc_ap.pdf',
        bbox_inches='tight',
        dpi=300
    )
    plt.show()
```

```python
plot_validation_metrics_combined(
    mlflow_client=mlflow_client,
    mapping=MAPPINGS_CROPPED,
    title="Porównanie metryk walidacji modeli korzystających z przycinania zdjęć"
)
```

```python
import matplotlib.pyplot as plt

def plot_train_loss_history(metric_history: list, title: str) -> None:

    steps = []
    values = []

    for entry in metric_history:
        step_value = entry.step
        metric_value = entry.value

        if not step_value in steps:
            steps.append(step_value)
            values.append(metric_value)

    plt.figure(figsize=(10, 6))
    plt.plot(steps, values, marker='o')
    plt.title(title)
    plt.xlabel("Step")
    plt.ylabel("Train Loss")
    plt.grid()
    plt.show()
```
