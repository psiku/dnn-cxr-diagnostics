```python
def _get_run_info(run_id: str, mlflow_client: mlflow.tracking.MlflowClient) -> None:
    run = mlflow_client.get_run(run_id)

    print("Run ID:", run.info.run_id)
    print("Experiment ID:", run.info.experiment_id)
    print("Start Time:", run.info.start_time)
    print("End Time:", run.info.end_time)
    print("Metrics:", run.data.metrics)
    print("Parameters:", run.data.params)
    print("Tags:", run.data.tags)
```

```python
train_loss_history = mlflow_client.get_metric_history(ID, "train/loss")

train_loss_history
```

```python
import matplotlib.pyplot as plt

def plot_train_loss_history(metric_history: list, title: str) -> None:

    steps = []
    values = []

    for entry in metric_history:
        step_value = entry.step
        metric_value = entry.value

        if not step_value in steps:
            steps.append(step_value)
            values.append(metric_value)

    plt.figure(figsize=(10, 6))
    plt.plot(steps, values, marker='o')
    plt.title(title)
    plt.xlabel("Step")
    plt.ylabel("Train Loss")
    plt.grid()
    plt.show()
```

```python
plot_train_loss_history(train_loss_history, "Train Loss History")
```

```python
ap_micro_history = mlflow_client.get_metric_history(ID, "val/ap_micro")
plot_train_loss_history(ap_micro_history, "Validation AP Micro History")
```
