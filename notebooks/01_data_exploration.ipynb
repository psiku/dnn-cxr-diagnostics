# 01 — Data Exploration

This notebook provides an initial exploratory data analysis (EDA) of the chest X-ray dataset.

**Goals:**
- Understand class distribution
- Visualise sample images
- Check image sizes and statistics

```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd

DATA_DIR = Path("../data/raw")
```

```python
# Count images per class
class_counts = {}
for class_dir in sorted(DATA_DIR.iterdir()):
    if class_dir.is_dir():
        n = len(list(class_dir.glob("*.png"))) + len(list(class_dir.glob("*.jpg")))
        class_counts[class_dir.name] = n

df_counts = pd.Series(class_counts, name="count").reset_index()
df_counts.columns = ["class", "count"]
df_counts
```

```python
# Bar chart of class distribution
df_counts.plot.bar(x="class", y="count", legend=False)
plt.title("Class Distribution")
plt.ylabel("Number of images")
plt.tight_layout()
plt.show()
```
